```python
# 1. Clone repo (skip if you already opened this notebook inside the cloned folder) & install dependencies
import os, subprocess, sys

REPO_URL = "https://github.com/ArafatAhmedMubinOfficial/NazrulLora.git"
if not os.path.exists("src/chat.py"):
    subprocess.run(["git", "clone", "-b", "unsensored", REPO_URL, "NazrulLora"], check=True)
    os.chdir("NazrulLora")

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "transformers>=5.5.0", "peft", "accelerate", "bitsandbytes", "safetensors", "sentencepiece"], check=True)
```

```python
# 2. Load Nazrul-4B Heretic (4-bit Gemma-4-E4B base + local LoRA adapter if trained)
from pathlib import Path
from src.chat import load_nazrul_model, generate

model, tok = load_nazrul_model(Path("outputs/nazrul-lora"))
print("Nazrul-4B loaded on:", model.device)
```

```python
# 3. Try prompts in Kazi Nazrul Islam's style
prompts = [
    "কাজী নজরুল ইসলামের ভঙ্গিতে অন্যায় ও পরাধীনতার বিরুদ্ধে জাগরণ ও বিদ্রোহের উপর একটি কবিতা লেখো।",
    "কাজী নজরুলের ভঙ্গিতে বর্ষা, মেঘ ও বিরহ নিয়ে একটি নজরুলগীতি রচনা করো।",
    "কাজী নজরুল ইসলামের ওজস্বী গদ্যশৈলীতে তরুণ সমাজ ও তারুণ্যের শক্তি নিয়ে একটি অনুচ্ছেদ লেখো।",
    "Write a passionate Bangla poem in the style of Kazi Nazrul Islam about freedom and youth."
]

for p in prompts:
    print("=" * 60)
    print("PROMPT:", p)
    print("-" * 60)
    print(generate(model, tok, p, max_new_tokens=240, temperature=0.72, rhyme_candidates=2))
```
